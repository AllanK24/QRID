# Yolov8 NO AUGMENTATION

## FULL PRECISION FP32

```python
import torch
from ultralytics import YOLO
from ultralytics.utils.benchmarks import benchmark

device = "cuda" if torch.cuda.is_available() else "cpu"
device
```

### Nano

```python
benchmark(model=YOLO("yolov8n.pt"), data="coco.yaml", imgsz=640, format="engine", device=device)

### The `benchmark` function automatically updates all dependencies to the latest versions needed for conversion to TensorRT (like ONNX).
### It also updates TensorRT to its latest version
### If dataset for benchmarking isn't provided, the `benchmark` function will download it itself
```

### Small

```python
benchmark(model=YOLO("yolov8s.pt"), data="coco.yaml", imgsz=640, format="engine", device=device)
```

### Medium

```python
benchmark(model=YOLO("yolov8m.pt"), data="coco.yaml", imgsz=640, format="engine", device=device)
```

### Large

```python
benchmark(model=YOLO("yolov8l.pt"), data="coco.yaml", imgsz=640, format="engine", device=device)
```

### Extra Large

```python
benchmark(model=YOLO("yolov8x.pt"), data="coco.yaml", imgsz=640, format="engine", device=device)
```
